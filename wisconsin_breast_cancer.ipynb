### 4.2 Duplicate analysis

"Duplicate" can mean different things depending on which columns we compare. We examine three distinct definitions:

1. **Exact duplicate rows** — rows where *all 11 columns* are identical.
2. **Repeated identifiers** — rows sharing the same `sample_code_number` (patient ID), which may represent repeat visits or data-entry reuse.
3. **Repeated predictive-feature profiles** — rows where the 9 predictive feature values are identical, regardless of the identifier or target.

These categories **overlap** and use different definitions, so their counts should never be added together.

In each case, `duplicated(keep='first')` counts only the *extra* occurrences (the first appearance is not counted as a duplicate).

```python
# 1. Exact duplicate rows (all 11 columns identical)
exact_dupes = df.duplicated(keep='first').sum()
print(f"Exact duplicate rows (extra occurrences): {exact_dupes}")

# 2. Repeated identifiers (same sample_code_number)
repeated_ids = df['sample_code_number'].duplicated(keep='first').sum()
print(f"Repeated identifiers (extra occurrences):  {repeated_ids}")

# 3. Repeated predictive-feature profiles (same 9 feature values)
repeated_profiles = df.duplicated(subset=feature_names, keep='first').sum()
print(f"Repeated feature profiles (extra occurrences): {repeated_profiles}")
```

**Interpretation:**

- **8 exact duplicate rows:** These are rows where all 11 values (including the identifier and target) are completely identical to an earlier row. These may represent data-entry errors or re-submissions.
- **54 repeated identifiers:** The same `sample_code_number` appears more than once. The dataset documentation notes that samples arrived in chronological batches, so the same patient could have multiple records collected at different times. A repeated identifier does not automatically indicate a data error.
- **236 repeated feature profiles:** When looking at only the 9 predictive features, 236 rows match an earlier row. This is a broader count because different patients can have identical cytological scores. Rows with the same feature values may have different identifiers or different target labels.

**Note:** Whether to remove any of these duplicates is a preprocessing decision and is deferred to Task 2 (Data Preprocessing) below.

# Task 2: Data Preprocessing

## 1. Objective

The objective of this task is to prepare the Breast Cancer Wisconsin (Original) dataset for supervised machine learning classification.

The preprocessing process includes:
- Preserving the original dataset.
- Identifying and removing exact duplicate records.
- Separating input features from the target variable.
- Excluding irrelevant features, such as the sample ID.
- Splitting the dataset into training and testing sets.
- Handling missing values.
- Applying feature standardization where necessary.
- Preventing data leakage by fitting preprocessing steps using the training data only.

The target variable represents two classes:
- `2`: Benign
- `4`: Malignant

For model evaluation, malignant cases will be treated as the positive class.


## 2. Create a Working Copy

The dataset was loaded and explored in Task 1. Therefore, it does not need to be loaded again as long as the original DataFrame `df` is still available in the current notebook session.

A separate working copy will be created to preserve the original dataset and allow preprocessing without modifying the original data.

```python
# Preserve the original dataset and create a working copy
df_original = df.copy()
df_clean = df.copy()
feature_columns = feature_names

print("Original dataset shape:", df_original.shape)
print("Working dataset shape:", df_clean.shape)
```


## 3. Duplicate Records

Exact duplicate records are rows in which all column values are identical.

Duplicate records may cause some observations to be counted more than once. Therefore, exact duplicates will be identified and removed from the working dataset.

Repeated sample IDs or repeated feature values are not automatically considered duplicates, because they may represent different records.

```python
# Count exact duplicate records before removal
duplicates_before = df_clean.duplicated().sum()
print("Exact duplicate records before removal:", duplicates_before)

# Remove exact duplicate records
df_clean = df_clean.drop_duplicates().copy()

# Check the dataset after removal
duplicates_after = df_clean.duplicated().sum()
print("Exact duplicate records after removal:", duplicates_after)
print("Dataset shape after removing duplicates:", df_clean.shape)
```

### Repeated Sample ID Investigation

Exact duplicate records were removed in the previous step. However, repeated sample IDs are a separate data quality issue because records sharing the same `sample_code_number` may have different feature values or target classes.

A further investigation identified:

- 39 repeated sample IDs.
- 85 records belonging to these repeated IDs.
- 4 sample IDs associated with multiple target classes.
- 39 repeated sample IDs with different feature profiles.

These records were not automatically removed because a repeated sample ID does not necessarily indicate an exact duplicate.

To prevent records sharing the same sample ID from appearing in both training and testing sets, a group-aware split was used. This resulted in zero shared sample IDs between the two subsets.

```python

# Investigate repeated sample IDs
id_counts = df_clean.groupby("sample_code_number").size()
repeated_ids = id_counts[id_counts > 1]

# Count distinct target classes for each sample ID
class_counts = df_clean.groupby("sample_code_number")["class"].nunique()
ids_with_multiple_classes = class_counts[class_counts > 1]

# Count distinct feature profiles for each sample ID
feature_profile_counts = df_clean.groupby("sample_code_number")[
    feature_columns
].apply(lambda group: group.drop_duplicates().shape[0])

ids_with_different_features = feature_profile_counts[
    (id_counts > 1) & (feature_profile_counts > 1)
]

print("Repeated sample IDs:", len(repeated_ids))
print("Records belonging to repeated IDs:", int(repeated_ids.sum()))
print("IDs with multiple target classes:", len(ids_with_multiple_classes))
print("Repeated IDs with different feature profiles:",
      len(ids_with_different_features))
```


## 4. Separate Features and Target

The dataset contains nine predictive features and one target variable.

The `sample_code_number` column is an identifier rather than a predictive feature, so it will be excluded from the model inputs.

The target variable is `class`, where:
- `2` represents benign cases.
- `4` represents malignant cases.

The target will be encoded as `0` for benign and `1` for malignant. This encoding allows malignant cases to be treated as the positive class during evaluation.

```python
# Separate the nine predictive features from the target
feature_columns = [
    "clump_thickness",
    "uniformity_of_cell_size",
    "uniformity_of_cell_shape",
    "marginal_adhesion",
    "single_epithelial_cell_size",
    "bare_nuclei",
    "bland_chromatin",
    "normal_nucleoli",
    "mitoses"
]

X = df_clean[feature_columns].copy()

# Encode the target: benign = 0, malignant = 1
y = df_clean["class"].map({2: 0, 4: 1})

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nTarget distribution:")
print(y.value_counts().sort_index().rename(index={0: "Benign", 1: "Malignant"}))
```


## 5. Group-Aware Train-Test Split

A group-aware split is used to prevent records sharing the same sample ID from appearing in both the training and testing sets.

StratifiedGroupKFold is used to maintain an approximately balanced class distribution while keeping each sample ID within a single subset.

The selected fold provides an approximately 80:20 training-to-testing split.

```python
from sklearn.model_selection import StratifiedGroupKFold

groups = df_clean["sample_code_number"]

# StratifiedGroupKFold ensures no patient sample ID appears in both train and test splits
sgkf = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)
train_idx, test_idx = next(sgkf.split(X, y, groups))

X_train, X_test = X.iloc[train_idx].copy(), X.iloc[test_idx].copy()
y_train, y_test = y.iloc[train_idx].copy(), y.iloc[test_idx].copy()

print("Training features shape:", X_train.shape)
print("Testing features shape:", X_test.shape)
print("Training target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)

print("\nTraining class distribution:")
print(y_train.value_counts(normalize=True).rename("Proportion").round(4))

print("\nTesting class distribution:")
print(y_test.value_counts(normalize=True).rename("Proportion").round(4))

# Check for sample IDs shared between training and testing
train_ids = set(groups.iloc[train_idx])
test_ids = set(groups.iloc[test_idx])

shared_ids = train_ids.intersection(test_ids)

print("\nNumber of shared sample IDs:", len(shared_ids))
print("Shared sample IDs:", sorted(shared_ids) if shared_ids else "None")
```


## 6. Handle Missing Values

During Task 1, missing values were identified in the `bare_nuclei` feature.

Missing values will be handled using median imputation. The median will be calculated from the training data only and then applied to both the training and testing sets.

This approach prevents information from the test set from influencing the preprocessing process.

```python
from sklearn.impute import SimpleImputer

# Fit the imputer on the training data only
imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

print("Missing values in training data after imputation:",
      int(np.isnan(X_train_imputed).sum()))

print("Missing values in testing data after imputation:",
      int(np.isnan(X_test_imputed).sum()))
```


## 7. Feature Standardization

Feature standardization transforms numerical features to have approximately zero mean and unit standard deviation.

This step is useful for distance-based and optimization-based models, such as KNN, SVM, and ANN, because features with larger scales may otherwise have an excessive influence on the learning process.

The scaler will be fitted using the training data only and then applied to the testing data to prevent data leakage.

Tree-based models, such as Decision Tree and Random Forest, generally do not require feature standardization.

```python
from sklearn.preprocessing import StandardScaler

# Fit the scaler on the training data only
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)

print("Scaled training data shape:", X_train_scaled.shape)
print("Scaled testing data shape:", X_test_scaled.shape)
```


## 8. Final Preprocessing Checks

The final step is to verify that:
- The training and testing sets have the expected dimensions.
- No missing values remain after imputation.
- The target contains only the encoded classes `0` and `1`.
- The training and testing sets are separate.
- The preprocessing steps were fitted using training data only.

```python
print("Final preprocessing summary")
print("-" * 40)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)

print("Missing values in training set:",
      int(np.isnan(X_train_imputed).sum()))

print("Missing values in testing set:",
      int(np.isnan(X_test_imputed).sum()))

print("Target classes:", sorted(y.unique()))

print("Training and testing indices overlap:",
      bool(set(X_train.index).intersection(set(X_test.index))))
```

## Preprocessing Summary

The Breast Cancer Wisconsin (Original) dataset was prepared for supervised machine learning classification.

The preprocessing steps included:

- Preserving the original dataset and creating a working copy.
- Removing 8 exact duplicate records, resulting in 691 records.
- Separating the 9 predictive features from the target variable.
- Mapping the original target labels (2 and 4) to binary labels (0 and 1), representing benign and malignant cases, respectively.
- Investigating repeated sample IDs and documenting records associated with multiple target classes.
- Applying a group-aware train-test split to prevent sample IDs from appearing in both subsets.
- Imputing missing feature values using the median calculated from the training data.
- Applying standardization using statistics learned from the training data only.
- Verifying the final dataset dimensions, target labels, missing values, and numerical validity.

The resulting training set contains 552 records, and the testing set contains 139 records, each with 9 predictive features.

The processed data is ready for the next stage of the machine learning workflow, subject to documenting and reviewing the identified sample ID inconsistencies.

```python
# Combine the processed training and testing data
X_train_final = pd.DataFrame(
    X_train_scaled,
    columns=feature_columns,
    index=X_train.index
)

X_test_final = pd.DataFrame(
    X_test_scaled,
    columns=feature_columns,
    index=X_test.index
)

X_final = pd.concat([X_train_final, X_test_final]).sort_index()

# Combine the target values in the same original row order
y_final = pd.concat([y_train, y_test]).sort_index()

# Create the final processed dataset
final_dataset = X_final.copy()
final_dataset["class"] = y_final

# Export to CSV
os.makedirs("data", exist_ok=True)
final_dataset.to_csv(os.path.join("data", "final_preprocessed_dataset.csv"), index=False)

print("Final dataset shape:", final_dataset.shape)
print("Missing values:", int(final_dataset.isna().sum().sum()))
print("File saved as: data/final_preprocessed_dataset.csv")
```

## 6. Boxplots and Outlier Analysis

Boxplots provide a five-number summary (minimum, first quartile $Q_1$, median, third quartile $Q_3$, and maximum) while flagging potential statistical outliers beyond the standard Tukey fences:
$$\text{Lower Fence} = Q_1 - 1.5 \times \text{IQR}, \quad \text{Upper Fence} = Q_3 + 1.5 \times \text{IQR}, \quad \text{IQR} = Q_3 - Q_1$$
We analyze distributions and outlier occurrences across both diagnostic classes.


```python
# Create comparative boxplots for all predictive features
fig, axes = plt.subplots(3, 3, figsize=(16, 13))
axes = axes.flatten()

for i, col in enumerate(feature_columns):
    ax = axes[i]
    sns.boxplot(
        data=df_vis,
        x="Diagnosis",
        y=col,
        hue="Diagnosis",
        legend=False,
        palette=PALETTE_CLASS,
        width=0.45,
        fliersize=5,
        flierprops={"marker": "o", "markerfacecolor": "#f39c12", "markeredgecolor": "#333333"},
        ax=ax
    )
    ax.set_title(f"{col.replace('_', ' ').title()}", fontsize=12, fontweight="bold", pad=8)
    ax.set_xlabel("Diagnostic Class", fontsize=10)
    ax.set_ylabel("Score (1–10)", fontsize=10)
    ax.set_yticks(range(1, 11))
    ax.grid(axis="y", linestyle="--", alpha=0.5)

plt.suptitle("Figure 3.4: Comparative Boxplots by Diagnostic Class and Outlier Detection (Tukey IQR Fences)", fontsize=15, fontweight="bold", y=1.01)
plt.tight_layout()
plt.show()

# Quantify outliers per feature and class using the standard Tukey 1.5 * IQR rule
outlier_records = []
for col in feature_columns:
    for cls_name, grp in df_vis.groupby("Diagnosis"):
        vals = grp[col].dropna()
        q1 = vals.quantile(0.25)
        med = vals.median()
        q3 = vals.quantile(0.75)
        iqr = q3 - q1
        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr
        outliers = vals[(vals < lower_bound) | (vals > upper_bound)]
        outlier_records.append({
            "Feature": col,
            "Diagnosis": cls_name,
            "Total N": len(vals),
            "Q1": q1,
            "Median": med,
            "Q3": q3,
            "IQR": iqr,
            "Lower Fence": lower_bound,
            "Upper Fence": upper_bound,
            "Outlier Count": len(outliers),
            "Outlier %": round((len(outliers) / len(vals)) * 100, 2),
            "Outlier Values": sorted(outliers.unique().tolist())
        })

outlier_summary = pd.DataFrame(outlier_records)
print("Statistical Outlier Summary Table (Tukey 1.5 * IQR Rule):")
display(outlier_summary[outlier_summary["Outlier Count"] > 0].reset_index(drop=True))
```

### Interpretation: Boxplots and Outlier Analysis

- **Outlier Quantification (Tukey IQR Method):**
  - Applying standard Tukey fences ($\text{Lower} = Q_1 - 1.5 \times \text{IQR}$, $\text{Upper} = Q_3 + 1.5 \times \text{IQR}$) identifies statistical outliers across the two cohorts:
    - **Benign Cohort ($n = 453$, $n = 439$ for `bare_nuclei`):**
      - `uniformity_of_cell_shape`: **107 outliers** (23.62%), values in [2, 3, 4, 5, 6, 7, 8].
      - `single_epithelial_cell_size`: **93 outliers** (20.53%), values in [1, 3, 4, 5, 6, 7, 8, 10].
      - `marginal_adhesion`: **83 outliers** (18.32%), values in [2, 3, 4, 5, 6, 9, 10].
      - `uniformity_of_cell_size`: **78 outliers** (17.22%), values in [2, 3, 4, 6, 7, 8, 9].
      - `bare_nuclei`: **57 outliers** (12.98%), values in [2.0, 3.0, 4.0, 5.0, 7.0, 8.0, 10.0].
      - `normal_nucleoli`: **56 outliers** (12.36%), values in [2, 3, 4, 5, 6, 7, 8, 9].
      - `mitoses`: **13 outliers** (2.87%), values in [2, 3, 5, 7, 8].
      - `bland_chromatin`: **7 outliers** (1.55%), value: [7].
      - `clump_thickness`: **0 outliers** (0.00%).
    - **Malignant Cohort ($n = 238$):**
      - `mitoses`: **29 outliers** (12.18%), values in [7, 8, 10].
      - All remaining eight features in the malignant cohort have **0 outliers** under Tukey's rule due to broader interquartile spreads ($\text{IQR} = 3$ to $7$).
- **Statistical Mechanism of Collapsed Interquartile Range ($IQR = 0$):**
  - An IQR outlier is a **univariate statistical classification**, not evidence of data entry error, sensor malfunction, or corrupted records.
  - In the benign cohort, several features have a concentration of values at a single score exceeding 75% of observations. Consequently, the 25th percentile ($Q_1$), the median ($Q_2$), and the 75th percentile ($Q_3$) all equal **1** (or all equal **2** for `single_epithelial_cell_size`).
  - When $Q_1 = Q_3$, the interquartile range collapses to **$\text{IQR} = 0.0$**. The resulting upper fence becomes:
    $$\text{Upper Fence} = Q_3 + 1.5 \times 0.0 = Q_3$$
    Under this strict statistical rule, any observation deviating from the central value (e.g., scores $\ge 2$ when $Q_3 = 1$, or scores $\ne 2$ when $Q_1 = Q_3 = 2$) is automatically designated as an upper outlier, regardless of whether that value represents a severe abnormality or merely minor variation.
- **Preservation Decision in Data Preprocessing:**
  - All observed values fall strictly within the valid, documented integer range of [1, 10].
  - Because these points represent valid empirical measurements within the documented attribute domains, removing or winsorizing them would artificially truncate natural variance and risk discarding informative signal.
  - Therefore, all observations flagged by Tukey's rule are **intentionally retained** without alteration for subsequent modeling.


## 7. Correlation Heatmap (Feature Relationships and Multicollinearity)

We examine linear correlations among the nine predictive features and their correlation with the binary diagnostic outcome (`class`: 0 = Benign, 1 = Malignant) to detect multicollinearity and determine feature associations with tumor malignancy. Formal two-tailed significance tests are conducted using `scipy.stats.pearsonr`.


```python
# Build correlation matrix including the binary-encoded target (0 = Benign, 1 = Malignant)
corr_df = df_clean[feature_columns].copy()
corr_df["malignant_class"] = df_clean["class"].map({2: 0, 4: 1})

corr_matrix = corr_df.corr(method="pearson")

fig, ax = plt.subplots(figsize=(11, 8.5))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="vlag",
    vmin=-1.0,
    vmax=1.0,
    center=0,
    square=True,
    linewidths=0.7,
    cbar_kws={"label": "Pearson Correlation Coefficient (r)", "shrink": 0.8},
    ax=ax
)

ax.set_title("Figure 3.5: Pearson Correlation Heatmap (Features and Malignancy Target, n = 691)", fontsize=14, fontweight="bold", pad=15)
ax.set_xticklabels([c.replace('_', ' ').title() for c in corr_matrix.columns], fontsize=9.5, rotation=45, ha="right")
ax.set_yticklabels([c.replace('_', ' ').title() for c in corr_matrix.columns], fontsize=9.5, rotation=0)

plt.tight_layout()
plt.show()

# Conduct formal two-tailed Pearson correlation significance tests (scipy.stats.pearsonr)
target_corr_records = []
for feat in feature_columns:
    # Drop missing bare_nuclei entries to perform valid pairwise test
    valid_data = corr_df[[feat, "malignant_class"]].dropna()
    r_val, p_val = stats.pearsonr(valid_data[feat], valid_data["malignant_class"])
    target_corr_records.append({
        "Feature": feat,
        "Pearson r": round(r_val, 4),
        "Rounded r": round(r_val, 2),
        "Two-tailed p-value": f"{p_val:.2e}",
        "Significant (p < 0.001)": p_val < 0.001
    })

target_corr_df = pd.DataFrame(target_corr_records).sort_values("Pearson r", ascending=False).reset_index(drop=True)
print("Ranked Pearson Correlation with Malignant Target (0=Benign, 1=Malignant):")
display(target_corr_df)

# Pairwise test between uniformity_of_cell_size and uniformity_of_cell_shape
r_ss, p_ss = stats.pearsonr(corr_df["uniformity_of_cell_size"], corr_df["uniformity_of_cell_shape"])
print(f"\nMulticollinearity Pairwise Test (Cell Size vs. Cell Shape):")
print(f"  r = {r_ss:.4f} (rounds to {round(r_ss, 2)}), two-tailed p = {p_ss:.2e} (Significant at p < 0.001)")
```

### Interpretation: Correlation Heatmap and Multicollinearity

- **Correlation vs. Causation:**
  - Pearson correlation coefficients quantify the degree of linear association between pairs of features; they do not establish biological or clinical causality. A strong positive correlation between an attribute (e.g., `bare_nuclei`) and tumor malignancy indicates that elevated values frequently co-occur with malignant diagnoses in this dataset, not that cellular atypia directly causes malignant transformation.
- **Multicollinearity and Feature Redundancy:**
  - An exceptionally strong positive correlation exists between **`uniformity_of_cell_size`** and **`uniformity_of_cell_shape`** ($r = 0.9054$, rounding to **$0.91$**; two-tailed test $p = 1.40 \times 10^{-258} < 0.001$).
  - Additional notable pairwise correlations include `uniformity_of_cell_size` with `bland_chromatin` ($r = 0.76$) and with `bare_nuclei` ($r = 0.69$).
  - When two or more features exhibit high collinearity, they convey overlapping, redundant information about underlying cellular morphology.
- **Impact of Correlated Features on Machine Learning Algorithms:**
  - **Distance-Based Algorithms (e.g., KNN):** In algorithms that rely on distance metrics such as Euclidean distance ($d(\mathbf{u}, \mathbf{v}) = \sqrt{\sum (u_i - v_i)^2}$), including multiple highly correlated features effectively gives disproportionate weight to the underlying trait they jointly represent. For example, retaining both cell size and cell shape may cause morphological uniformity to dominate distance calculations relative to independent features such as `mitoses`. Feature selection or dimensionality reduction (e.g., PCA) may be considered if redundancy degrades neighborhood search.
  - **Linear Models (e.g., Logistic Regression):** Severe collinearity inflates parameter variance, making individual coefficient estimates unstable and sensitive to minor data fluctuations, although regularized formulations (L1/L2 penalty) can mitigate this effect.
  - **Kernel Methods (SVM) & Tree Ensembles (Random Forest):** Non-linear kernel SVMs and decision-tree ensembles are generally less sensitive to linear collinearity, as orthogonal decision splits evaluate features sequentially. However, feature importance rankings in random forests can be dispersed across correlated surrogates.
  - **Empirical Modeling Takeaway:** High correlation does not automatically benefit or harm classification models. The practical effect on predictive performance depends on the algorithm, the regularization applied, and the validation strategy, which will be evaluated empirically on the testing set in Tasks 4 and 5.
- **Association with Diagnostic Target (`malignant_class`) & Consistent p-values:**
  - Formal two-tailed hypothesis testing confirms that all nine predictive features exhibit positive associations with tumor malignancy that are statistically significant at **$p < 0.001$**, with p-values matching the executed statistical test table:
    - **`bare_nuclei`**: $r = 0.8205 \approx 0.82$ ($p = 1.25 \times 10^{-165}$)

  - **Tree-Based Ensembles (Decision Trees, Random Forests):** Because orthogonal split finding depends only on the monotonic order of feature values, tree models are invariant to monotonic transformations and perform identically on raw and standardized inputs.

## 3. Cross-Validation Protocol and Scoring Setup

- **Inner CV:** `StratifiedGroupKFold(n_splits=5)` on the 552 training records, grouped by `sample_code_number`.
- **Scoring:** malignant (`1`) is the positive class. `GridSearchCV` selects hyperparameters by the mean **F2-score** across folds (`refit="f2"`) and also records Recall, Precision, F1, ROC-AUC, and Accuracy.
- `return_train_score=True` lets us compare training vs. validation scores to detect overfitting.

```python
cv = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

# zero_division=0: a model that predicts no malignant case gets precision/F-scores of 0 (no warning spam)
scoring = {
    "f2": make_scorer(fbeta_score, beta=2, zero_division=0),
    "recall": make_scorer(recall_score, zero_division=0),
    "precision": make_scorer(precision_score, zero_division=0),
    "f1": make_scorer(f1_score, zero_division=0),
    "roc_auc": "roc_auc",
    "accuracy": "accuracy",
}
METRICS = list(scoring.keys())

# Inspect the inner folds
fold_rows = []
for k, (tr, va) in enumerate(cv.split(X_train, y_train, groups_train), start=1):
    fold_rows.append({
        "Fold": k,
        "Train size": len(tr),
        "Validation size": len(va),
        "Validation malignant %": round(100 * y_train.iloc[va].mean(), 2),
        "Shared IDs": len(set(groups_train.iloc[tr]) & set(groups_train.iloc[va])),
    })
display(pd.DataFrame(fold_rows).set_index("Fold"))
```

### Interpretation: Inner Cross-Validation Folds

- Each of the 5 folds trains on **441–443** records and validates on **109–111** records.
- The malignant share in every validation fold stays between **33.94% and 35.14%**, close to the overall training share of 34.42%, so every fold is representative of the class imbalance.
- **0 shared IDs** between the training and validation part of every fold: repeated sample IDs never appear on both sides, which keeps the validation scores honest.

### 3.1 Helper functions

`tune_model` runs the grid search strictly on `(X_train, y_train)` and returns the fitted search object; `summarize_best` reports the cross-validated metrics (mean ± std) of the selected configuration and its train–validation gap.

```python
def tune_model(name, pipeline, param_grid):
    # Exhaustive grid search with group-aware stratified CV on training data only, refit on F2-score.
    search = GridSearchCV(
        estimator=pipeline,
        param_grid=param_grid,
        scoring=scoring,
        refit="f2",
        cv=cv,
        n_jobs=N_JOBS,
        return_train_score=True,
    )
    start = time.perf_counter()
    search.fit(X_train, y_train, groups=groups_train)
    elapsed = time.perf_counter() - start
    n_candidates = len(search.cv_results_["params"])
    print(f"{name}: evaluated {n_candidates} hyperparameter combinations x {cv.get_n_splits()} folds "
          f"= {n_candidates * cv.get_n_splits()} fits in {elapsed:.1f} s")
    print("Best hyperparameters:")
    for p, v in search.best_params_.items():
        print(f"  - {p.replace('model__', '')}: {v}")
    return search


def summarize_best(name, search):
    # Mean +/- std of each CV metric for the best configuration, plus train score for overfitting checks.
    i = search.best_index_
    res = search.cv_results_
    row = {"Model": name}
    for m in METRICS:
        row[f"CV {m} mean"] = res[f"mean_test_{m}"][i]
        row[f"CV {m} std"] = res[f"std_test_{m}"][i]
        row[f"Train {m} mean"] = res[f"mean_train_{m}"][i]
    table = pd.DataFrame({
        "Validation (mean)": [res[f"mean_test_{m}"][i] for m in METRICS],
        "Validation (std)": [res[f"std_test_{m}"][i] for m in METRICS],
        "Training (mean)": [res[f"mean_train_{m}"][i] for m in METRICS],
    }, index=[m.upper() if m != "roc_auc" else "ROC-AUC" for m in METRICS])
    table["Train - Val gap"] = table["Training (mean)"] - table["Validation (mean)"]
    display(table.round(4))
    return row


cv_summary_rows = []
searches = {}
```

## 5. K-Nearest Neighbors (KNN)

KNN classifies a sample by a majority vote of its *k* closest training samples. Because it relies on Euclidean/Manhattan distances, features are **standardized** inside the pipeline (fitted strictly on each CV training fold).

**Hyperparameter grid**

| Hyperparameter | Values | Role |
|:---|:---|:---|
| `n_neighbors` (k) | 1, 3, 5, …, 31 | Small k → flexible (high variance); large k → smoother (high bias). Odd values avoid ties. |
| `weights` | `uniform`, `distance` | Equal votes vs. closer neighbours weigh more. |
| `p` | 1 (Manhattan), 2 (Euclidean) | Distance metric. |

```python
knn_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", KNeighborsClassifier()),
])
knn_grid = {
    "model__n_neighbors": list(range(1, 32, 2)),
    "model__weights": ["uniform", "distance"],
    "model__p": [1, 2],
}
searches["KNN"] = tune_model("KNN", knn_pipeline, knn_grid)
cv_summary_rows.append(summarize_best("KNN", searches["KNN"]))
```

```python
knn_res = pd.DataFrame(searches["KNN"].cv_results_)
best = searches["KNN"].best_params_

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
styles = {("uniform", 1): ("#2b7bba", "-"), ("uniform", 2): ("#2b7bba", "--"),
          ("distance", 1): ("#e67e22", "-"), ("distance", 2): ("#e67e22", "--")}
for (w, p), grp in knn_res.groupby(["param_model__weights", "param_model__p"]):
    color, ls = styles[(w, p)]
    axes[0].plot(grp["param_model__n_neighbors"], grp["mean_test_f2"], marker="o", ms=4,
                 color=color, linestyle=ls, label=f"weights={w}, p={p}")
axes[0].axvline(best["model__n_neighbors"], color="#d9534f", linestyle=":", label=f"Selected k = {best['model__n_neighbors']}")
axes[0].set_title("A. Validation F2-score vs. k (all weight/metric combinations)", fontweight="bold")
axes[0].set_xlabel("Number of neighbours (k)")
axes[0].set_ylabel("Mean CV F2-score")
axes[0].legend(fontsize=9)

sel = knn_res[(knn_res["param_model__weights"] == best["model__weights"]) & (knn_res["param_model__p"] == best["model__p"])]
axes[1].plot(sel["param_model__n_neighbors"], sel["mean_train_f2"], marker="o", ms=4, color="#27ae60", label="Training F2")
axes[1].plot(sel["param_model__n_neighbors"], sel["mean_test_f2"], marker="o", ms=4, color="#8e44ad", label="Validation F2")
axes[1].fill_between(sel["param_model__n_neighbors"].astype(float),
                     sel["mean_test_f2"] - sel["std_test_f2"], sel["mean_test_f2"] + sel["std_test_f2"],
                     color="#8e44ad", alpha=0.15, label="Validation ± 1 std")
axes[1].set_title(f"B. Training vs. validation F2 (weights={best['model__weights']}, p={best['model__p']})", fontweight="bold")
axes[1].set_xlabel("Number of neighbours (k)")
axes[1].set_ylabel("Mean F2-score")
axes[1].legend(fontsize=9)

plt.suptitle("Figure 4.1: KNN Hyperparameter Tuning (5-fold Stratified Group CV)", fontsize=14, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```

### Interpretation: KNN Tuning

- **Selected configuration:** `n_neighbors = 5`, `weights = distance`, `p = 2` (Euclidean), with a mean CV F2 of **0.9583 ± 0.0261** (Recall **0.9636**, Precision **0.9403**, ROC-AUC **0.9795**).
- **Effect of k (Figure 4.1A):** `k = 1` performs poorly for every variant (F2 ≈ 0.92) and has the widest fold-to-fold spread (high variance). | `min_samples_leaf` | 1, 2, 5, 10, 20 | Minimum samples per leaf (smooths decisions). |
| `class_weight` | None, `balanced` | `balanced` up-weights the minority (malignant) class. |

```python
dt_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", DecisionTreeClassifier(random_state=RANDOM_STATE)),
])
dt_grid = {
    "model__criterion": ["gini", "entropy"],
    "model__max_depth": [2, 3, 4, 5, 6, 8, 10, None],
    "model__min_samples_leaf": [1, 2, 5, 10, 20],
    "model__class_weight": [None, "balanced"],
}
searches["Decision Tree"] = tune_model("Decision Tree", dt_pipeline, dt_grid)
cv_summary_rows.append(summarize_best("Decision Tree", searches["Decision Tree"]))
```

```python
dt_res = pd.DataFrame(searches["Decision Tree"].cv_results_)
best = searches["Decision Tree"].best_params_
depth_order = [2, 3, 4, 5, 6, 8, 10, None]
depth_labels = [str(d) if d is not None else "None" for d in depth_order]

# Panel A: depth curve with the other hyperparameters fixed at their best values
mask = ((dt_res["param_model__criterion"] == best["model__criterion"]) &
        (dt_res["param_model__min_samples_leaf"] == best["model__min_samples_leaf"]) &
        (dt_res["param_model__class_weight"].astype(str) == str(best["model__class_weight"])))
sel = dt_res[mask].copy()
sel["depth_label"] = sel["param_model__max_depth"].apply(lambda d: "None" if d is None or (isinstance(d, float) and np.isnan(d)) else str(int(d)))
sel = sel.set_index("depth_label").loc[depth_labels]

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
x = np.arange(len(depth_labels))
axes[0].plot(x, sel["mean_train_f2"], marker="o", color="#27ae60", label="Training F2")
axes[0].plot(x, sel["mean_test_f2"], marker="o", color="#e67e22", label="Validation F2")
axes[0].fill_between(x, sel["mean_test_f2"] - sel["std_test_f2"], sel["mean_test_f2"] + sel["std_test_f2"],
                     color="#e67e22", alpha=0.15, label="Validation ± 1 std")
axes[0].set_xticks(x, depth_labels)
axes[0].set_title(f"A. F2 vs. max_depth (criterion={best['model__criterion']}, "
                  f"min_samples_leaf={best['model__min_samples_leaf']}, class_weight={best['model__class_weight']})",
                  fontweight="bold", fontsize=10.5)
axes[0].set_xlabel("max_depth")
axes[0].set_ylabel("Mean F2-score")
axes[0].legend(fontsize=9)

# Panel B: heatmap of validation F2 for max_depth x min_samples_leaf (best criterion & class_weight)
mask_b = ((dt_res["param_model__criterion"] == best["model__criterion"]) &
          (dt_res["param_model__class_weight"].astype(str) == str(best["model__class_weight"])))
heat = dt_res[mask_b].copy()
heat["max_depth"] = heat["param_model__max_depth"].apply(lambda d: "None" if d is None or (isinstance(d, float) and np.isnan(d)) else str(int(d)))
heat = heat.pivot_table(index="max_depth", columns="param_model__min_samples_leaf", values="mean_test_f2").loc[depth_labels]
sns.heatmap(heat, annot=True, fmt=".3f", cmap="Oranges", ax=axes[1], cbar_kws={"label": "Mean CV F2"})
axes[1].set_title("B. Validation F2: max_depth x min_samples_leaf", fontweight="bold")
axes[1].set_xlabel("min_samples_leaf")
axes[1].set_ylabel("max_depth")

plt.suptitle("Figure 4.2: Decision Tree Hyperparameter Tuning", fontsize=14, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```

- **Complexity control works (Figure 4.2):** with `min_samples_leaf = 1`, deep trees (`max_depth ≥ 8` or unlimited) fall to a validation F2 of **0.866–0.870**, a clear sign of overfitting. - **Small train–validation gap:** training F2 is 0.9649 vs. 0.9516 in validation (gap **0.0133**), so the shallow tree generalises well.
- **Recall vs. precision trade-off:** the balanced class weights give the tree a high recall (**0.9737**) but the lowest precision of all models (**0.8754**), i.e. it flags more benign samples as malignant. A depth-3 tree has at most 8 leaves, which also makes it the most interpretable model.

## 7. Random Forest (RF)

A random forest builds an ensemble of decision trees, each trained on a bootstrap sample with a random subset of features considered at each split. As with DT, **no scaling** is needed.

| `max_depth` | None, 3, 5, 8 | Depth limit per tree. |
| `max_features` | `sqrt`, 0.5 | Features considered per split (decorrelates trees). |
| `min_samples_leaf` | 1, 3, 5 | Minimum samples per leaf. |
| `class_weight` | None, `balanced` | Minority-class re-weighting. |

```python
rf_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", RandomForestClassifier(random_state=RANDOM_STATE)),
])
rf_grid = {
    "model__n_estimators": [100, 300],
    "model__max_depth": [None, 3, 5, 8],
    "model__max_features": ["sqrt", 0.5],
    "model__min_samples_leaf": [1, 3, 5],
    "model__class_weight": [None, "balanced"],
}
searches["Random Forest"] = tune_model("Random Forest", rf_pipeline, rf_grid)
cv_summary_rows.append(summarize_best("Random Forest", searches["Random Forest"]))
```

All of them use `max_features = sqrt` and `class_weight = balanced`, so these two settings matter most, while the number of trees (100 vs. 300) and depth limits make practically no difference. Among tied configurations, `GridSearchCV` keeps the first one in rank order.

## 8. Artificial Neural Network (ANN — Multilayer Perceptron)

An `MLPClassifier` is a feed-forward neural network trained with back-propagation (Adam optimiser). Gradient-based training is sensitive to feature scale, so features are **standardized** inside the pipeline. With 552 training samples, compact networks with L2 regularisation (`alpha`) are explored to control capacity.

| Hyperparameter | Values | Role |
|:---|:---|:---|
| `hidden_layer_sizes` | (8,), (16,), (32,), (16, 8), (32, 16) | Network architecture (one or two hidden layers). |
| `activation` | `relu`, `tanh` | Hidden-layer non-linearity. |
| `alpha` | 0.0001, 0.001, 0.01, 0.1, 1.0 | L2 regularisation strength. |
| `learning_rate_init` | 0.001, 0.01 | Adam step size. |

`max_iter=2000` gives the optimiser enough epochs to converge on this dataset.

```python
ann_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", MLPClassifier(max_iter=2000, random_state=RANDOM_STATE)),
])
ann_grid = {
    "model__hidden_layer_sizes": [(8,), (16,), (32,), (16, 8), (32, 16)],
    "model__activation": ["relu", "tanh"],
    "model__alpha": [0.0001, 0.001, 0.01, 0.1, 1.0],
    "model__learning_rate_init": [0.001, 0.01],
}
searches["ANN (MLP)"] = tune_model("ANN (MLP)", ann_pipeline, ann_grid)
cv_summary_rows.append(summarize_best("ANN (MLP)", searches["ANN (MLP)"]))
```

```python
ann_res = pd.DataFrame(searches["ANN (MLP)"].cv_results_)
best = searches["ANN (MLP)"].best_params_
ann_res["architecture"] = ann_res["param_model__hidden_layer_sizes"].astype(str)

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
sel = ann_res[(ann_res["param_model__activation"] == best["model__activation"]) &
              (ann_res["param_model__learning_rate_init"] == best["model__learning_rate_init"])]
arch_colors = dict(zip(sel["architecture"].unique(), sns.color_palette("tab10", sel["architecture"].nunique())))
for arch, grp in sel.groupby("architecture"):
    grp = grp.sort_values("param_model__alpha")
    axes[0].plot(grp["param_model__alpha"].astype(float), grp["mean_test_f2"], marker="o", ms=4,
                 color=arch_colors[arch], label=f"hidden layers = {arch}")
axes[0].set_xscale("log")
axes[0].set_title(f"A. Validation F2 vs. L2 penalty alpha (activation={best['model__activation']}, "
                  f"lr={best['model__learning_rate_init']})", fontweight="bold", fontsize=10.5)
axes[0].set_xlabel("alpha (log scale)")
axes[0].set_ylabel("Mean CV F2-score")
axes[0].legend(fontsize=9)

best_mlp = searches["ANN (MLP)"].best_estimator_.named_steps["model"]
axes[1].plot(best_mlp.loss_curve_, color="#8e44ad")
axes[1].set_title(f"B. Training loss curve of the selected network ({best_mlp.n_iter_} epochs)", fontweight="bold")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Log-loss (training)")

plt.suptitle("Figure 4.4: ANN (MLP) Hyperparameter Tuning", fontsize=14, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
print(f"Selected network converged: {best_mlp.n_iter_ < best_mlp.max_iter} "
      f"(epochs used: {best_mlp.n_iter_} of max {best_mlp.max_iter}); final training loss = {best_mlp.loss_:.4f}")
```

### Interpretation: ANN (MLP) Tuning

- **Selected configuration:** one hidden layer with **8 neurons**, `activation = relu`, `alpha = 0.0001`, `learning_rate_init = 0.001`, with mean CV F2 **0.9657 ± 0.0220** and the **highest ROC-AUC of all models (0.9920 ± 0.0037)**.

The larger networks (32,) and (32, 16) score lower (0.942–0.958) and get closer to the best score mainly under strong regularisation (`alpha = 1.0`), which shows that extra capacity is not needed for 9 features and 552 samples.

- **Convergence (Figure 4.4B):** the loss decreases smoothly and the optimiser stopped after **275 of 2000** epochs because the loss stopped improving, ending at a training log-loss of **0.0926**.

## 9. Support Vector Machine (SVM)

An SVM finds the maximum-margin boundary between classes; the RBF kernel allows non-linear boundaries. Margins are distance-based, so features are **standardized** inside the pipeline. Two sub-grids are searched: a linear kernel and an RBF kernel.

| Hyperparameter | Values | Role |
|:---|:---|:---|
| `kernel` | `linear`, `rbf` | Linear vs. non-linear decision boundary. |
| `C` | 0.001 – 1000 (log grid) | Trade-off between margin width and training errors (small C = stronger regularisation). |
| `gamma` (RBF only) | `scale`, 0.001, 0.01, 0.1, 1 | Kernel width (large gamma = very local, wiggly boundary). |
| `class_weight` | None, `balanced` | Minority-class re-weighting. |

ROC-AUC is computed from the SVM decision function, so `probability=True` is not required.

```python
svm_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", SVC(random_state=RANDOM_STATE)),
])
svm_grid = [
    {
        "model__kernel": ["linear"],
        "model__C": [0.001, 0.01, 0.1, 1, 10, 100],
        "model__class_weight": [None, "balanced"],
    },
    {
        "model__kernel": ["rbf"],
        "model__C": [0.01, 0.1, 1, 10, 100, 1000],
        "model__gamma": ["scale", 0.001, 0.01, 0.1, 1],
        "model__class_weight": [None, "balanced"],
    },
]
searches["SVM"] = tune_model("SVM", svm_pipeline, svm_grid)
cv_summary_rows.append(summarize_best("SVM", searches["SVM"]))
```

```python
svm_res = pd.DataFrame(searches["SVM"].cv_results_)
best = searches["SVM"].best_params_
best_cw = str(best["model__class_weight"])

fig, axes = plt.subplots(1, 2, figsize=(16, 5.5))
rbf = svm_res[(svm_res["param_model__kernel"] == "rbf") &
              (svm_res["param_model__class_weight"].astype(str) == best_cw)].copy()
rbf["gamma"] = rbf["param_model__gamma"].astype(str)
heat = rbf.pivot_table(index="gamma", columns="param_model__C", values="mean_test_f2")
heat = heat.loc[["scale", "0.001", "0.01", "0.1", "1"]]
# Colour scale starts at 0.85 so differences among good settings stay visible (lower values appear lightest)
sns.heatmap(heat, annot=True, fmt=".3f", cmap="Reds", vmin=0.85, ax=axes[0], cbar_kws={"label": "Mean CV F2"})
axes[0].set_title(f"A. RBF kernel: validation F2 for C x gamma (class_weight={best_cw})", fontweight="bold")
axes[0].set_xlabel("C")
axes[0].set_ylabel("gamma")

lin = svm_res[svm_res["param_model__kernel"] == "linear"]
for cw, grp in lin.groupby(lin["param_model__class_weight"].astype(str)):
    grp = grp.sort_values("param_model__C")
    axes[1].plot(grp["param_model__C"].astype(float), grp["mean_test_f2"], marker="o",
                 label=f"linear, class_weight={cw}", color="#2b7bba" if cw == "None" else "#d9534f")
axes[1].set_xscale("log")
axes[1].set_title("B. Linear kernel: validation F2 vs. C", fontweight="bold")
axes[1].set_xlabel("C (log scale)")
axes[1].set_ylabel("Mean CV F2-score")
axes[1].legend(fontsize=9)

plt.suptitle("Figure 4.5: SVM Hyperparameter Tuning", fontsize=14, fontweight="bold", y=1.02)
plt.tight_layout()
plt.show()
```

### Interpretation: SVM Tuning

- **Selected configuration:** RBF kernel with `C = 1`, `gamma = scale`, `class_weight = balanced`, giving mean CV F2 **0.9670 ± 0.0246**, the second-highest recall (**0.9794**), and ROC-AUC **0.9833**.
- **RBF heat map (Figure 4.5A):** `gamma = scale` equals 1 / (n_features × variance) ≈ 1/9 ≈ 0.11 for standardized data, so it behaves almost exactly like `gamma = 0.1` (both 0.967 at C = 1). Very strong regularisation with a very smooth kernel (`C = 0.01`, `gamma = 0.001`) collapses towards predicting the majority class (F2 = **0.291**), while large C with mid-range gamma (e.g. `C ≥ 100`, `gamma = scale/0.1`) reduces F2 to ≈ 0.92 by fitting the training noise.
- **Linear kernel (Figure 4.5B):** the best linear SVM (`C = 0.1`, balanced) reaches F2 ≈ **0.960**, only slightly below the RBF kernel. The classes are therefore almost linearly separable in this feature space. `class_weight = balanced` improves the linear SVM at every C.
- **Generalisation:** train F2 0.9758 vs. CV F2 0.9670 (gap **0.0088**).

## 10. Empirical Cross-Validated Comparison & Model Selection

All five algorithms were tuned on the exact same training set (`n = 552`) using the same 5 group-aware folds, without any prior assumption favouring one model over another. Below we compare their experimental validation results to select the leading candidate, while keeping the testing set (`n = 139`) untouched for Task 5.

```python
cv_summary = pd.DataFrame(cv_summary_rows).set_index("Model")

display_cols = {}
for m in METRICS:
    label = "ROC-AUC" if m == "roc_auc" else m.upper() if m in ("f2", "f1") else m.capitalize()
    display_cols[label] = cv_summary[f"CV {m} mean"].map("{:.4f}".format) + " ± " + cv_summary[f"CV {m} std"].map("{:.4f}".format)
comparison_table = pd.DataFrame(display_cols)
print("Cross-validated performance of the tuned models (mean ± std over 5 group-aware folds):")
display(comparison_table)

best_cv_model = cv_summary.drop(index="Baseline (Majority)")["CV f2 mean"].idxmax()
print(f"\nLeading model selected empirically by mean CV F2-score: {best_cv_model} "
      f"(CV F2 = {cv_summary.loc[best_cv_model, 'CV f2 mean']:.4f} ± {cv_summary.loc[best_cv_model, 'CV f2 std']:.4f})")

gap_table = pd.DataFrame({
    "Train F2": cv_summary["Train f2 mean"],
    "CV F2": cv_summary["CV f2 mean"],
    "Gap (Train - CV)": cv_summary["Train f2 mean"] - cv_summary["CV f2 mean"],
}).drop(index="Baseline (Majority)")
print("\nOverfitting check (F2-score):")
display(gap_table.round(4))
```

```python
plot_metrics = ["f2", "recall", "precision", "roc_auc"]
metric_titles = {"f2": "F2-score (selection metric)", "recall": "Recall (Sensitivity)",
                 "precision": "Precision", "roc_auc": "ROC-AUC"}
models_to_plot = [m for m in cv_summary.index if m != "Baseline (Majority)"]

fig, axes = plt.subplots(1, 4, figsize=(20, 5.5), sharey=False)
for ax, m in zip(axes, plot_metrics):
    means = cv_summary.loc[models_to_plot, f"CV {m} mean"]
    stds = cv_summary.loc[models_to_plot, f"CV {m} std"]
    ax.bar(range(len(models_to_plot)), means, yerr=stds, capsize=5,
           color=[MODEL_COLORS[n] for n in models_to_plot], edgecolor="#2c3e50")
    for i, (mu, sd) in enumerate(zip(means, stds)):
        ax.text(i, mu + sd + 0.003, f"{mu:.3f}", ha="center", va="bottom", fontsize=9)
    ax.set_xticks(range(len(models_to_plot)), models_to_plot, rotation=30, ha="right")
    lo = max(0.0, (means - stds).min() - 0.03)
    ax.set_ylim(lo, max(1.01, (means + stds).max() + 0.015))
    ax.set_title(metric_titles[m], fontweight="bold")
    ax.set_ylabel("Mean CV score (± 1 std)")
plt.suptitle("Figure 4.6: Cross-Validated Performance of the Five Tuned Models (training set, 5-fold group-aware CV)",
             fontsize=14, fontweight="bold", y=1.03)
plt.tight_layout()
plt.show()
```

### Interpretation: Empirical Cross-Validated Model Selection

- **Evidence-based selection (no prior assumptions):** No algorithm was assumed superior before experimentation. All five candidate families—distance-based (KNN), single tree (DT), ensemble bagging (RF), neural network (ANN), and kernel margin (SVM)—were tuned and compared empirically on the exact same 5 group-aware folds.
- **All five tuned models beat the majority baseline by a wide margin:** CV F2 ranges from **0.9516 to 0.9741** (vs. 0.0000 for the baseline), and ROC-AUC ranges from **0.9702 to 0.9920** (vs. 0.5000).
- **Best model by experimental CV results — Random Forest:** achieves the highest selection metric (**F2 = 0.9741 ± 0.0125**), highest recall (**0.9843**), highest F1 (**0.9594**), highest accuracy (**0.9711**), and lowest fold-to-fold variability (F2 std **0.0125**).
- **Close contenders:** **SVM** ranks second on F2 (**0.9670**) and recall (**0.9794**), while **ANN (MLP)** achieves the highest ranking discrimination (**ROC-AUC = 0.9920**) with virtually zero train–validation gap.
- **KNN** reaches the highest precision (**0.9403**) but the lowest recall (**0.9636**) of the five, missing slightly more malignant cases. **Decision Tree** scores lowest on F2 (**0.9516**) and precision (**0.8754**), trading precision for simplicity (depth 3, ≤ 8 leaves).
- **Statistical note for Task 5:** Because the top three models (RF, SVM, ANN) lie within ~0.008 F2 of one another—smaller than one fold standard deviation (0.0125–0.0246)—all five refitted pipelines are carried forward to Task 5 for final out-of-sample evaluation on the untouched 139-sample test set.

## 11. Final Hyperparameters, Persisting Models & Leakage Sanity Check

`GridSearchCV(refit="f2")` has already re-trained each best configuration exclusively on the **552 training records**. We record the final hyperparameters, persist all five fitted pipelines alongside `data/train_test_split.csv`, and run a programmatic sanity check confirming that preprocessing was fitted strictly on `X_train` (`n = 552`) and that `X_test` (`n = 139`) was never touched.

```python
final_params = pd.DataFrame({
    name: {"Best hyperparameters": ", ".join(f"{k.replace('model__', '')}={v}" for k, v in s.best_params_.items()),
           "CV F2 (mean)": round(s.best_score_, 4),
           "Combinations searched": len(s.cv_results_["params"])}
    for name, s in searches.items()
}).T
display(final_params)

os.makedirs("models", exist_ok=True)
file_names = {"KNN": "knn", "Decision Tree": "decision_tree", "Random Forest": "random_forest",
              "ANN (MLP)": "ann_mlp", "SVM": "svm"}
for name, s in searches.items():
    path = os.path.join("models", f"{file_names[name]}_pipeline.joblib")
    joblib.dump(s.best_estimator_, path)
    print(f"Saved {name:<14} -> {path}")

# Save the exact train/test split (raw, un-imputed features + sample IDs) for Task 5
split_df = pd.concat([
    df_clean.loc[X_train.index, ["sample_code_number"] + feature_columns].assign(**{"class": y_train, "split": "train"}),
    df_clean.loc[X_test.index, ["sample_code_number"] + feature_columns].assign(**{"class": y_test, "split": "test"}),
]).sort_index()
split_df.to_csv(os.path.join("data", "train_test_split.csv"), index_label="row_index")
cv_summary.to_csv(os.path.join("models", "cv_results_summary.csv"))
print(f"\nSaved split -> data/train_test_split.csv  ({(split_df['split'] == 'train').sum()} train / {(split_df['split'] == 'test').sum()} test)")
print("Saved CV summary -> models/cv_results_summary.csv")
```

```python
# Sanity check: verify saved pipeline reproduction and strict training-only preprocessing fit
reloaded = joblib.load(os.path.join("models", "svm_pipeline.joblib"))
same = (reloaded.predict(X_train) == searches["SVM"].best_estimator_.predict(X_train)).all()

scaler_samples = {
    name: int(searches[name].best_estimator_.named_steps["scaler"].n_samples_seen_)
    for name in ["KNN", "ANN (MLP)", "SVM"]
}
assert all(n == len(X_train) for n in scaler_samples.values())

print("Reloaded SVM pipeline reproduces in-memory predictions:", bool(same))
print(f"Samples seen by StandardScaler inside pipelines (must equal len(X_train) = {len(X_train)}):", scaler_samples)
print("Test set (n = 139) used anywhere during preprocessing, tuning, or model selection: NO (reserved for Task 5)")
```

### Interpretation: Saved Artefacts & Methodological Verification

- The five fitted pipelines (imputer → scaler where required → tuned classifier), refitted exclusively on the **552 training records**, are stored in `models/`, and reloading them reproduces the in-memory predictions identically.
- `n_samples_seen_ = 552` inside every fitted `StandardScaler` confirms programmatically that preprocessing parameters were learned strictly from `X_train` (`n = 552`) and never exposed to `X_test` (`n = 139`).
- `data/train_test_split.csv` preserves the exact 552 train / 139 test partition with `sample_code_number` so Task 5 can evaluate the saved pipelines on the completely unseen test set.

# Task 5: Model Evaluation

## 1. Objective and Evaluation Protocol (Malignant = Positive Class)

In Task 4 the five algorithms (KNN, Decision Tree, Random Forest, ANN, SVM) were tuned **using the training set only** (5-fold group-aware cross-validation). In this task each tuned pipeline is evaluated **once** on the **held-out test set** (139 records, 20% of the data) that was never used for preprocessing, tuning, or model selection.

**Protocol**

- **Positive class = Malignant (`1`)**, negative class = Benign (`0`), as required.
- The saved pipelines (`models/*.joblib`) contain imputation, scaling (where needed) and the model, all fitted on the training data only, so raw test features can be passed in directly with no leakage.
- Test data is used for **reporting**, not for changing hyperparameters.
## 2. Explanation of the Evaluation Metrics

Let **TP** = malignant correctly flagged, **FN** = malignant missed (predicted benign), **FP** = benign wrongly flagged malignant, **TN** = benign correctly cleared.

| Metric | Formula | Meaning in this project |
|:---|:---|:---|
| **Accuracy** | (TP + TN) / (TP + TN + FP + FN) | Share of all patients classified correctly. Can look good even when the minority class is handled poorly (here only ~35% are malignant). |
| **Precision** | TP / (TP + FP) | Of the cases flagged malignant, how many truly are. Low precision means unnecessary biopsies and patient anxiety (false alarms). |
| **Recall (sensitivity)** | TP / (TP + FN) | Of all truly malignant cases, how many are caught. **The most critical metric clinically**, because a missed cancer (FN) is the most harmful error. |
| **F1-score** | 2 · Precision · Recall / (Precision + Recall) | Harmonic mean of precision and recall; one number that is low if either one is low. |
| **ROC-AUC** | Area under the TPR-vs-FPR curve over all thresholds | Threshold-independent ability to rank malignant cases above benign ones (0.5 = random, 1.0 = perfect). |
| **Confusion matrix** | 2 × 2 table of TP, FP, FN, TN | Shows exactly *which* kind of mistakes each model makes. |


## 3. Loading the Test Set and the Trained Models


```python
from sklearn.metrics import accuracy_score, roc_auc_score, roc_curve, confusion_matrix

POSITIVE_CLASS = 1          # 1 = Malignant (positive class), 0 = Benign
CLASS_NAMES = ["Benign", "Malignant"]

MODEL_FILES = {
    "KNN": "knn", "Decision Tree": "decision_tree", "Random Forest": "random_forest",
    "ANN (MLP)": "ann_mlp", "SVM": "svm",
}

# ---- 1. Load the exact train/test split saved in Task 4 (raw, un-imputed features) ----
split_df = pd.read_csv(os.path.join("data", "train_test_split.csv"))
feature_cols = [c for c in split_df.columns
                if c not in ("row_index", "sample_code_number", "class", "split")]

train_part = split_df[split_df["split"] == "train"]
test_part = split_df[split_df["split"] == "test"]

X_eval = test_part[feature_cols]
y_eval = test_part["class"].astype(int)

# ---- 2. Integrity checks on the held-out test set ----
shared_ids = set(train_part["sample_code_number"]) & set(test_part["sample_code_number"])
assert len(train_part) == 552 and len(test_part) == 139
assert len(shared_ids) == 0, "Patient IDs leak between train and test"
assert set(test_part["row_index"]) == set(X_test.index), "Test rows differ from Task 2 split"

print(f"Held-out test set : {X_eval.shape[0]} records x {X_eval.shape[1]} features")
print(f"Malignant (positive) : {int(y_eval.sum())}  ({y_eval.mean():.1%})")
print(f"Benign (negative)    : {int((1 - y_eval).sum())}  ({1 - y_eval.mean():.1%})")
print(f"Missing raw values in test features (handled by each pipeline's imputer): {int(X_eval.isna().sum().sum())}")
print(f"Patient IDs shared with the training set: {len(shared_ids)}")

# ---- 3. Load the five tuned pipelines saved in Task 4 (imputer -> scaler -> model, fitted on training data only) ----
models = {name: joblib.load(os.path.join("models", f"{f}_pipeline.joblib")) for name, f in MODEL_FILES.items()}
for name, m in models.items():
    print(f"Loaded {name:<14} -> steps: {[s for s, _ in m.steps]}")


def get_scores(model, X):
    """Continuous malignancy score used for ROC-AUC: P(malignant) if available, else the SVM margin."""
    if hasattr(model, "predict_proba"):
        return model.predict_proba(X)[:, 1]
    return model.decision_function(X)
```

## 4. Evaluation of All Models: Accuracy, Precision, Recall, F1-score, ROC-AUC (One Comparison Table)


```python
# Predict once on the untouched test set and compute every required metric (malignant = positive class)
preds, scores, rows = {}, {}, []
for name, model in models.items():
    preds[name] = model.predict(X_eval)
    scores[name] = get_scores(model, X_eval)
    tn, fp, fn, tp = confusion_matrix(y_eval, preds[name], labels=[0, 1]).ravel()
    rows.append({
        "Model": name,
        "Accuracy": accuracy_score(y_eval, preds[name]),
        "Precision": precision_score(y_eval, preds[name], pos_label=POSITIVE_CLASS, zero_division=0),
        "Recall": recall_score(y_eval, preds[name], pos_label=POSITIVE_CLASS),
        "F1-score": f1_score(y_eval, preds[name], pos_label=POSITIVE_CLASS),
        "ROC-AUC": roc_auc_score(y_eval, scores[name]),
        "TP": tp, "FP": fp, "FN": fn, "TN": tn,
    })

results = pd.DataFrame(rows).set_index("Model")
METRIC_COLS = ["Accuracy", "Precision", "Recall", "F1-score", "ROC-AUC"]

print("Test-set performance of all five tuned models (n = 139, positive class = Malignant):\n")
display(results.round(4))

print("\nBest value per metric:")
for m in METRIC_COLS:
    best = results[m].max()
    winners = ", ".join(results.index[results[m] == best])
    print(f"  {m:<10} {best:.4f}  ->  {winners}")

os.makedirs("models", exist_ok=True)
results.round(6).to_csv(os.path.join("models", "test_results_summary.csv"))
print("\nSaved -> models/test_results_summary.csv")
```

### Interpretation: Comparison Table

- **Recall:** Random Forest, ANN and SVM all reach **1.000**: they detect all 48 malignant test cases. KNN misses 1 (0.979) and the Decision Tree misses 2 (0.958).
- **Precision / false alarms:** KNN is the most precise (0.979, only 1 false alarm), followed by ANN (0.960), Random Forest (0.941) and SVM (0.906). The Decision Tree is clearly the weakest (0.807; 11 benign cases flagged as malignant).
- **F1-score:** ANN (**0.980**) and KNN (0.979) are practically tied at the top, then Random Forest (0.970), SVM (0.951) and the Decision Tree (0.876).
- **ROC-AUC:** ANN and SVM are the highest (0.9998), then Random Forest (0.998), KNN (0.987) and the Decision Tree (0.966). All five models rank malignant above benign cases very well, so the SVM's lower precision comes from its decision threshold, not from poor ranking.
- **Accuracy:** KNN and ANN share the top (0.986, 2 errors out of 139).

## 7. Model Comparison and Best Overall Model

**Selection rule** (stated before looking at the ranking, based on the clinical objective):

1. **Highest recall** for the malignant class, because missing a cancer is the costliest error;
2. ties broken by **F1-score**, which accounts for false alarms;
3. remaining ties broken by **ROC-AUC**.


```python
# Transparent selection rule (decided from the clinical objective, then applied to the test results):
#   1) highest Recall  (a missed malignant case is the most costly error)
#   2) then highest F1-score (balances recall with precision / false alarms)
#   3) then highest ROC-AUC (threshold-independent ranking quality)
ranking = results[METRIC_COLS].sort_values(["Recall", "F1-score", "ROC-AUC"], ascending=False)
ranking.insert(0, "Rank", range(1, len(ranking) + 1))
print("Models ranked by the rule  Recall -> F1-score -> ROC-AUC:")
display(ranking.round(4))

best_model = ranking.index[0]
print(f"\nBest overall model on the test set: {best_model}")
```

- **Highest ROC-AUC** (0.9998, tied with SVM) on the test set, and also the highest ROC-AUC in cross-validation (0.992), so this is not just a test-set coincidence.
